```python

def total_spent(purchases):
    total = 0                      # the notepad, set to 0 ONCE, before the loop
    for purchase in purchases:
        total = total + purchase   # add each receipt to the notepad
    return total                   # after the loop, hand back the notepad

purchases = [12.50, 8.00, 23.75, 5.25]
print(total_spent(purchases))
```

```python
def total_steps(days):
    total = 0
    for day in days:
        total = total + day
    return total

days = [8200, 10500, 6300, 12000, 9100]
print(total_steps(days))
```

```python
def count_passing(scores):
    count = 0
    for score in scores:
        if score >= 60:
            count += 1
    return (count)

scores = [85, 42, 91, 67, 58, 73]
print(count_passing(scores))
```

```python
def total_calories(meals):
    total = 0
    for meal in meals:
        total = meal + total
    return (total)

meals = [450, 600, 300, 250]
print(total_calories(meals))
```

```python
def total_distance(runs):
    total_miles = 0
    for run in runs:
        total_miles = run + total_miles
    return (total_miles)

runs = [3.1, 5.0, 2.5, 6.2, 4.0]
print(total_distance(runs))
```

```python
def highest_sale(sales):
    best_sale = 0
    for sale in sales:
        if best_sale < sale:
            best_sale = sale 
    return(str(best_sale) + " is the highest sale")

sales = [340, 210, 580, 195, 420]
print(highest_sale(sales))
```

```python
def lowest_score(scores):
    low_score = 8888
    for score in scores:
        if low_score > score:
            low_score = score
    return(str(low_score) + " is the lowest score")

scores =  [88, 72, 95, 61, 80]
print(lowest_score(scores))
```
