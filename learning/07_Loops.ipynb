```python
averages = [27.4, 19.8, 31.2, 22.1, 25.5]
line = 24.5

for avg in averages:
    if avg > line:
        print(avg, 'Over')
    else:
        print(avg, 'Under')
```

```python
players = [
    {'name': 'Luka', 'avg': 29.1, 'line': 28.5},
    {'name': 'Jokic', 'avg': 26.8, 'line': 27.5},
    {'name': 'Edwards', 'avg': 25.9, 'line': 24.5},
]

for player in players:
    if player['avg'] > player['line']:
        print(player['name'], 'Over')
    else:
        print(player['name'], 'Under')
```

```python
def analyze_slate(players):
  

    for player in players:
        if player['avg'] > player['line']:
            print(player['name'], 'Over')
        else:
            print(player['name'], 'Under')

analyze_slate(players)
```

```python
tonight = [
    {'name': 'Luka', 'avg': 29.1, 'line': 28.5},
    {'name': 'Tatum', 'avg': 27.2, 'line': 28.5},
]

analyze_slate(tonight)   # hand tonight's slate to the function
```

```python
def add(a, b):
    return a + b

result = add(3, 4)   # result now holds 7
print(result)        # 7
```

```python
def analyze_slate_two(players):
    results = []
    for player in players:
        if player['avg'] > player['line']:
            results.append(player['name'] + " Over")
        else:
            results.append(player['name'] + " Under")
    return results

props = analyze_slate_two(tonight)
print(props)
```

```python
props = analyze_slate_two(tonight)

print("Total props analyzed:", len(props))   # count them
```

```python
def analyze_player(name, avg):
    line = 24.5
    if avg > line:
        return (name + " OVER")
    else:
        return (name + " UNDER")

result = analyze_player('Luka', 29.1)
print(result)

print(analyze_player('Tatum', 22.1))
```

```python
slate = [('Luka', 29.1), ('Jokic', 22.8), ('Edwards', 26.5)]
for name, avg in slate:
    print(analyze_player(name, avg))
    
    # name is 'Luka', avg is 29.1 on the first loop
```

```python
def grade_workout(minutes):
    if minutes >= 60:
        return "Beast mode"
    elif minutes >= 30 and minutes <= 59:
        return "Solid session"
    else: 
        return "Get back in there"

print(grade_workout(29))
print(grade_workout(75))   # predict?
print(grade_workout(45))   # predict?
print(grade_workout(15))   # predict?
```
